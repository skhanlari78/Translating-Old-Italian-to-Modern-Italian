# Output File Generation for Translation Evaluation

This notebook processes the translation output CSV files from different models (MADLAD, NLLB, PHI-4, QWEN3) and generates JSONL files in the required format:
- Translation files: `Saffron_Mafia-hw2_transl-{modelname}.jsonl`
- LLM-as-judge files: `Saffron_Mafia-hw2_transl-judge-{modelname}_with_context.jsonl` and `Saffron_Mafia-hw2_transl-judge-{modelname}_without_context.jsonl`

For PHI-4 and QWEN3, separate files are generated for each model variant (0, 1, 2, 3) and approach (zero-shot, few-shot).

All files are organized in the `outputs` folder with separate subfolders for each model group:
- `outputs/madlad/`
- `outputs/nllb/`
- `outputs/phi-4/`
- `outputs/qwen3/`

```python
import pandas as pd
import json
import os
from pathlib import Path

# Create output directory structure
output_dir = Path('outputs')
output_dir.mkdir(exist_ok=True)

# Create subdirectories for each model
model_dirs = ['madlad', 'nllb', 'phi-4', 'qwen3']
for model_dir in model_dirs:
    (output_dir / model_dir).mkdir(exist_ok=True)

print("Created output directory structure:")
for model_dir in model_dirs:
    print(f"  outputs/{model_dir}/")
```

```python
def create_translation_entry(row, translation_col, model_name=None, variant=None):
    """Create a translation entry for JSONL format"""
    entry = {
        "id": f"{row.name}",  # Use row index as ID
        "author": row['Author'],
        "date": row['Date'],
        "region": row['Region'],
        "original_sentence": row['Sentence'],
        "translated_sentence": row[translation_col]
    }
    return entry

def create_judge_entry(row, translation_col, eval_col, variant=None):
    """Create a judge evaluation entry for JSONL format"""
    entry = {
        "id": f"{row.name}",  # Use row index as ID
        "author": row['Author'],
        "date": row['Date'],
        "region": row['Region'],
        "original_sentence": row['Sentence'],
        "translated_sentence": row[translation_col],
        "evaluation_score": row[eval_col]
    }
    return entry

def save_jsonl(data, filename, model_subfolder):
    """Save data to JSONL format in the appropriate model subfolder"""
    filepath = output_dir / model_subfolder / filename
    with open(filepath, 'w', encoding='utf-8') as f:
        for entry in data:
            f.write(json.dumps(entry, ensure_ascii=False) + '\n')
    print(f"Saved {len(data)} entries to {filepath}")
```

```python
# Process MADLAD files
print("Processing MADLAD files...")

# Load MADLAD data
madlad_translations = pd.read_csv('madlad/madlad_translations.csv')
madlad_with_eval = pd.read_csv('madlad/madlad_translations_with_eval.csv')

# Generate translation JSONL for MADLAD
madlad_transl_data = []
for idx, row in madlad_translations.iterrows():
    entry = create_translation_entry(row, 'Modern_Italian')
    madlad_transl_data.append(entry)

save_jsonl(madlad_transl_data, 'Saffron_Mafia-hw2_transl-madlad.jsonl', 'madlad')

# Generate judge JSONL for MADLAD - with context
madlad_judge_with_context_data = []
for idx, row in madlad_with_eval.iterrows():
    entry = create_judge_entry(row, 'Modern_Italian', 'gemini_eval')
    madlad_judge_with_context_data.append(entry)

save_jsonl(madlad_judge_with_context_data, 'Saffron_Mafia-hw2_transl-judge-madlad_with_context.jsonl', 'madlad')

# Generate judge JSONL for MADLAD - without context
madlad_judge_no_context_data = []
for idx, row in madlad_with_eval.iterrows():
    entry = create_judge_entry(row, 'Modern_Italian', 'gemini_eval_no_context')
    madlad_judge_no_context_data.append(entry)

save_jsonl(madlad_judge_no_context_data, 'Saffron_Mafia-hw2_transl-judge-madlad_without_context.jsonl', 'madlad')
```

```python
# Process NLLB files
print("\nProcessing NLLB files...")

# Load NLLB data
nllb_translations = pd.read_csv('nllb/nllb_translations.csv')
nllb_with_eval = pd.read_csv('nllb/nllb_translations_with_eval.csv')

# Generate translation JSONL for NLLB
nllb_transl_data = []
for idx, row in nllb_translations.iterrows():
    entry = create_translation_entry(row, 'Modern_Italian')
    nllb_transl_data.append(entry)

save_jsonl(nllb_transl_data, 'Saffron_Mafia-hw2_transl-nllb.jsonl', 'nllb')

# Generate judge JSONL for NLLB - with context
nllb_judge_with_context_data = []
for idx, row in nllb_with_eval.iterrows():
    entry = create_judge_entry(row, 'Modern_Italian', 'gemini_eval')
    nllb_judge_with_context_data.append(entry)

save_jsonl(nllb_judge_with_context_data, 'Saffron_Mafia-hw2_transl-judge-nllb_with_context.jsonl', 'nllb')

# Generate judge JSONL for NLLB - without context
nllb_judge_no_context_data = []
for idx, row in nllb_with_eval.iterrows():
    entry = create_judge_entry(row, 'Modern_Italian', 'gemini_eval_no_context')
    nllb_judge_no_context_data.append(entry)

save_jsonl(nllb_judge_no_context_data, 'Saffron_Mafia-hw2_transl-judge-nllb_without_context.jsonl', 'nllb')
```

```python
# Process PHI-4 files
print("\nProcessing PHI-4 files...")

# Load PHI-4 data
phi4_zeroshot = pd.read_csv('phi-4/phi-4_translations.csv')
phi4_zeroshot_eval = pd.read_csv('phi-4/phi-4_translations_with_eval.csv')
phi4_fewshot = pd.read_csv('phi-4/phi-4_translations_fewshot.csv')
phi4_fewshot_eval = pd.read_csv('phi-4/phi-4_translations_fewshot_with_eval.csv')

# Process variants 0-3 for zero-shot
for variant in range(4):
    translation_col = f'Modern_Italian_{variant}'
    
    # Translation JSONL for zero-shot
    phi4_zs_transl_data = []
    for idx, row in phi4_zeroshot.iterrows():
        if pd.notna(row[translation_col]):
            entry = create_translation_entry(row, translation_col)
            phi4_zs_transl_data.append(entry)
    
    save_jsonl(phi4_zs_transl_data, f'Saffron_Mafia-hw2_transl-phi-4-zeroshot-v{variant}.jsonl', 'phi-4')
    
    # Judge JSONL for zero-shot - with context
    eval_col = f'Modern_Italian_{variant}_gemini_eval'
    eval_no_context_col = f'Modern_Italian_{variant}_gemini_eval_no_context'
    
    phi4_zs_judge_with_context_data = []
    for idx, row in phi4_zeroshot_eval.iterrows():
        if pd.notna(row[translation_col]) and eval_col in row:
            if pd.notna(row[eval_col]):
                entry = create_judge_entry(row, translation_col, eval_col)
                phi4_zs_judge_with_context_data.append(entry)
    
    save_jsonl(phi4_zs_judge_with_context_data, f'Saffron_Mafia-hw2_transl-judge-phi-4-zeroshot-v{variant}_with_context.jsonl', 'phi-4')
    
    # Judge JSONL for zero-shot - without context
    phi4_zs_judge_no_context_data = []
    for idx, row in phi4_zeroshot_eval.iterrows():
        if pd.notna(row[translation_col]) and eval_no_context_col in row:
            if pd.notna(row[eval_no_context_col]):
                entry = create_judge_entry(row, translation_col, eval_no_context_col)
                phi4_zs_judge_no_context_data.append(entry)
    
    save_jsonl(phi4_zs_judge_no_context_data, f'Saffron_Mafia-hw2_transl-judge-phi-4-zeroshot-v{variant}_without_context.jsonl', 'phi-4')

# Process variants 0-3 for few-shot
for variant in range(4):
    translation_col = f'Modern_Italian_{variant}'
    
    # Translation JSONL for few-shot
    phi4_fs_transl_data = []
    for idx, row in phi4_fewshot.iterrows():
        if pd.notna(row[translation_col]):
            entry = create_translation_entry(row, translation_col)
            phi4_fs_transl_data.append(entry)
    
    save_jsonl(phi4_fs_transl_data, f'Saffron_Mafia-hw2_transl-phi-4-fewshot-v{variant}.jsonl', 'phi-4')
    
    # Judge JSONL for few-shot - with context
    eval_col = f'Modern_Italian_{variant}_gemini_eval'
    eval_no_context_col = f'Modern_Italian_{variant}_gemini_eval_no_context'
    
    phi4_fs_judge_with_context_data = []
    for idx, row in phi4_fewshot_eval.iterrows():
        if pd.notna(row[translation_col]) and eval_col in row:
            if pd.notna(row[eval_col]):
                entry = create_judge_entry(row, translation_col, eval_col)
                phi4_fs_judge_with_context_data.append(entry)
    
    save_jsonl(phi4_fs_judge_with_context_data, f'Saffron_Mafia-hw2_transl-judge-phi-4-fewshot-v{variant}_with_context.jsonl', 'phi-4')
    
    # Judge JSONL for few-shot - without context
    phi4_fs_judge_no_context_data = []
    for idx, row in phi4_fewshot_eval.iterrows():
        if pd.notna(row[translation_col]) and eval_no_context_col in row:
            if pd.notna(row[eval_no_context_col]):
                entry = create_judge_entry(row, translation_col, eval_no_context_col)
                phi4_fs_judge_no_context_data.append(entry)
    
    save_jsonl(phi4_fs_judge_no_context_data, f'Saffron_Mafia-hw2_transl-judge-phi-4-fewshot-v{variant}_without_context.jsonl', 'phi-4')
```

```python
# Process QWEN3 files
print("\nProcessing QWEN3 files...")

# Load QWEN3 data
qwen3_zeroshot = pd.read_csv('qwen3/qwen3_translations.csv')
qwen3_zeroshot_eval = pd.read_csv('qwen3/qwen3_translations_with_eval.csv')
qwen3_fewshot = pd.read_csv('qwen3/qwen3_translations_fewshot.csv')
qwen3_fewshot_eval = pd.read_csv('qwen3/qwen3_translations_fewshot_with_eval.csv')

# Process variants 0-3 for zero-shot
for variant in range(4):
    translation_col = f'Modern_Italian_{variant}'
    
    # Translation JSONL for zero-shot
    qwen3_zs_transl_data = []
    for idx, row in qwen3_zeroshot.iterrows():
        if pd.notna(row[translation_col]):
            entry = create_translation_entry(row, translation_col)
            qwen3_zs_transl_data.append(entry)
    
    save_jsonl(qwen3_zs_transl_data, f'Saffron_Mafia-hw2_transl-qwen3-zeroshot-v{variant}.jsonl', 'qwen3')
    
    # Judge JSONL for zero-shot - with context
    eval_col = f'Modern_Italian_{variant}_gemini_eval'
    eval_no_context_col = f'Modern_Italian_{variant}_gemini_eval_no_context'
    
    qwen3_zs_judge_with_context_data = []
    for idx, row in qwen3_zeroshot_eval.iterrows():
        if pd.notna(row[translation_col]) and eval_col in row:
            if pd.notna(row[eval_col]):
                entry = create_judge_entry(row, translation_col, eval_col)
                qwen3_zs_judge_with_context_data.append(entry)
    
    save_jsonl(qwen3_zs_judge_with_context_data, f'Saffron_Mafia-hw2_transl-judge-qwen3-zeroshot-v{variant}_with_context.jsonl', 'qwen3')
    
    # Judge JSONL for zero-shot - without context
    qwen3_zs_judge_no_context_data = []
    for idx, row in qwen3_zeroshot_eval.iterrows():
        if pd.notna(row[translation_col]) and eval_no_context_col in row:
            if pd.notna(row[eval_no_context_col]):
                entry = create_judge_entry(row, translation_col, eval_no_context_col)
                qwen3_zs_judge_no_context_data.append(entry)
    
    save_jsonl(qwen3_zs_judge_no_context_data, f'Saffron_Mafia-hw2_transl-judge-qwen3-zeroshot-v{variant}_without_context.jsonl', 'qwen3')

# Process variants 0-3 for few-shot
for variant in range(4):
    translation_col = f'Modern_Italian_{variant}'
    
    # Translation JSONL for few-shot
    qwen3_fs_transl_data = []
    for idx, row in qwen3_fewshot.iterrows():
        if pd.notna(row[translation_col]):
            entry = create_translation_entry(row, translation_col)
            qwen3_fs_transl_data.append(entry)
    
    save_jsonl(qwen3_fs_transl_data, f'Saffron_Mafia-hw2_transl-qwen3-fewshot-v{variant}.jsonl', 'qwen3')
    
    # Judge JSONL for few-shot - with context
    eval_col = f'Modern_Italian_{variant}_gemini_eval'
    eval_no_context_col = f'Modern_Italian_{variant}_gemini_eval_no_context'
    
    qwen3_fs_judge_with_context_data = []
    for idx, row in qwen3_fewshot_eval.iterrows():
        if pd.notna(row[translation_col]) and eval_col in row:
            if pd.notna(row[eval_col]):
                entry = create_judge_entry(row, translation_col, eval_col)
                qwen3_fs_judge_with_context_data.append(entry)
    
    save_jsonl(qwen3_fs_judge_with_context_data, f'Saffron_Mafia-hw2_transl-judge-qwen3-fewshot-v{variant}_with_context.jsonl', 'qwen3')
    
    # Judge JSONL for few-shot - without context
    qwen3_fs_judge_no_context_data = []
    for idx, row in qwen3_fewshot_eval.iterrows():
        if pd.notna(row[translation_col]) and eval_no_context_col in row:
            if pd.notna(row[eval_no_context_col]):
                entry = create_judge_entry(row, translation_col, eval_no_context_col)
                qwen3_fs_judge_no_context_data.append(entry)
    
    save_jsonl(qwen3_fs_judge_no_context_data, f'Saffron_Mafia-hw2_transl-judge-qwen3-fewshot-v{variant}_without_context.jsonl', 'qwen3')
```

```python
# Summary of generated files
print("\n" + "="*50)
print("SUMMARY OF GENERATED FILES")
print("="*50)

# List all generated JSONL files in the outputs directory
total_files = 0
for model_dir in model_dirs:
    model_path = output_dir / model_dir
    generated_files = [f for f in os.listdir(model_path) if f.startswith('Saffron_Mafia-hw2_transl') and f.endswith('.jsonl')]
    generated_files.sort()
    
    print(f"\n{model_dir.upper()} Model ({len(generated_files)} files):")
    print("-" * 40)
    
    for file in generated_files:
        file_path = model_path / file
        file_size = os.path.getsize(file_path)
        with open(file_path, 'r', encoding='utf-8') as f:
            line_count = sum(1 for line in f)
        
        file_type = "JUDGE" if "judge" in file else "TRANSLATION"
        criteria = ""
        if "with_context" in file:
            criteria = " (with context)"
        elif "without_context" in file:
            criteria = " (without context)"
        
        print(f"  {file_type:12}{criteria:16} | {file:50} | {line_count:3} entries | {file_size:6} bytes")
    
    total_files += len(generated_files)

print(f"\n" + "="*50)
print(f"Total files generated: {total_files}")
print("All files have been successfully generated in the outputs/ directory!")
```
